```python
import pandas as pd

p = pd.read_excel(r"C:\Users\LENOVO\Desktop\Comp Bio Project\data\raw\Phenotypes.xlsx", index_col=0)
print(p.shape)

for c in p.columns:
    vals = [str(v).lower() for v in p[c].dropna().unique()]
    if any(("basal" in v) or ("lum" in v) for v in vals):
        print("Found subtype labels in:", c)
        print(p[c].value_counts(dropna=False), "\n")
```
